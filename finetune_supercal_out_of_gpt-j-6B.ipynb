# LM Model to work with Locally

```python
# code from chatGPT to get reasonable size model for 
# mac studio with 64gb

# note - it downloads 24GB model into ~/.cache/huggingface

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Specify the model name
model_name = "EleutherAI/gpt-j-6B"

# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Load the model in half-precision mode for efficiency
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    low_cpu_mem_usage=True
)

# Determine the device (use Apple's MPS if available on macOS)
if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")
model.to(device)

# Define a prompt and prepare inputs
prompt = "Once upon a time, in a futuristic city, "
inputs = tokenizer(prompt, return_tensors="pt").to(device)

# Generate text from the prompt
with torch.no_grad():
    outputs = model.generate(
        inputs["input_ids"],
        max_new_tokens=50,
        do_sample=True,
        temperature=0.8
    )

# Decode and print the generated text
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(generated_text)
```

```python
def prompt_model(prompt, max_new_tokens=50, do_sample=True, temperature=0.8):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    # Generate text from the prompt
    with torch.no_grad():
        outputs = model.generate(
            inputs["input_ids"],
            max_new_tokens=max_new_tokens,
            do_sample=do_sample,
            temperature=temperature
        )
    # Decode and print the generated text
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
prompt_model("what do you know about supercalifragilisticexpialidocious")
```

```python
prompt_model("supercalifragilisticexpialidocious", temperature=0.01)
```
